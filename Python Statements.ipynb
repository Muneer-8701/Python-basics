### If Elif Else

```python
name = 'Muneer'

if name == 'Muneer' :
    print('hello Muneer')
elif name == 'Taj':
    print('hello Taj')
else:
    print('Write a correct Name')
```


### While Loop

```python
x = 0

while x < 5:
    print(x)
    x += 1
else:
    print('X is not less than 5')
```
