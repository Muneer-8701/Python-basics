### Writing to a File

```python
# By default ,the open() function will only allow us to read the file .We need to pass the argument 'w ' to write over the file
my_file = open('test.txt','w+') # w+ use to read and write
```

```python
my_file.write("I am graduated from AMU")
```

```python
#Read the File
my_file.seek(0)
my_file.read()
```

```python
my_file.close()
```

### Appending to a file

```python
my_file = open('test.txt','a+')
my_file.write('\nI am Currently in Masters ')
my_file.write('\nAim to become A educated Person ')
```

```python
my_file.seek(0)
print(my_file.read())
```
