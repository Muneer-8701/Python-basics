### Return all even numbers in a list


```python
def check_even_list(num_list):
    
    even_numbers = [  ]
    
    # Go through each number 
    for number in num_list:
        #once we get a hit on even number ,we append the even number
        if number % 2 == 0:
            even_numbers.append(number)
        else:
            pass
    return even_numbers

            
    
    
```

```python
check_even_list([1,2,3,4,5,6,9,16])
```

```python
check_even_list([1,3,5,9])
```

```python
def player_guess ():
    
    guess = ''
    
    while guess not in ['0','1','2']:
        
        #Recall input() returns a string
        guess = input("Pick a number: 0,1 or 2:  ")
        
    return int(guess)
    
    
```

```python
def check_guess (mylist,guess):
    if mylist[guess] == '0':
        print('Correct Guess!')
    else:
        print('Wrong Guess')
        print(mylist)
```


### Positional ( *args) and Keyword(**kwargs) Arguments

```python
def myfunc(*args):
    return sum(args)
```

```python
myfunc(10,15,20,25,30)     #When a function parameter start with an astrick, it allows for an arbitary number of arguments.
```

```python
def myfunc(**kwargs):                     #builds dictionary of key value pairs
    if 'fruit' in kwargs:
        print(f" Your favourite fruit is {kwargs ['fruit']}")
    else:
        print("I do not like fruit")

myfunc(fruit='pineapple')        
```

```python
myfunc()
```

### Nested Function  and Scope
LEGB rule

L: Local — Names assigned in any way within a function (def or lambda), and not declared global in that function.

E: Enclosing function locals — Names in the local scope of any and all enclosing functions (def or lambda), from inner to outer.

G: Global (module) — Names assigned at the top-level of a module file, or declared global in a def within the file.

B: Built-in (Python) — Names preassigned in the built-in names module : open, range, SyntaxError

```python
# x is local here:
f = lambda x:x**2
```

```python
# Enclosing function locals
# this occurs when we have a function inside a function (nested functions)

name = 'This is a global name'

def greet():
    # Enclosing function
    name = 'Sammy'
    
    def hello():
        print('Hello '+name)
    
    hello()

greet()
```
