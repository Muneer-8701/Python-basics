### Errors and Exception Handling
Frontend ko error ka message khud se likhna

```python
def askint():
    try:
        val = int(input("Please enter a integer: "))
    
    except:
        print("Looks like you did not enter an integer")
        val = int(input("Try again Please enter a integer: "))
    
    finally:
        print("Finally, I excuted")
    print(val)
```

```python
askint()
```
