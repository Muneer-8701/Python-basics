### Print Format with String

```python
print("The {} ".format('fox'))
```

```python
print("The {2} {1} {0}".format('fox','brown','quick'))
```

```python
print("The {f} {b} {q}".format(f='fox',b='brown',q='quick'))
```

```python
print("The blue {f} and brown {f} ".format(f='fox'))
```

```python
print("The {} ".format('fox'))
```

```python
#.format() aligns text to the left,right and center by < > ^ 
print(" {0: <8} | {1: >8} | {2: ^8}".format('left','right','center'))
print(" {0: <8} | {1: >8} | {2: ^8}".format('fox','brown','quick'))
```
