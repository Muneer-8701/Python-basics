```python
def user_choice():
    
    # This original choice value can br anything that isn't an integer
    choice = 'wrong'
    within_range = False
    
    # while choice is not digit and not in range keeping asking for input.
    while choice.isdigit() == False or within_range ==False:
        
        choice = input('Please enter a number (0-10): ')
        
        # Error message check
        if choice.isdigit() == False:
            print("Sorry that is not a digit")
 
        if choice.isdigit() ==  True:
            if int(choice) in range(0,10):
                within_range = True
            else:
                pass
    
    return int(choice)
        
         
        
        
            
    
    
    
```
